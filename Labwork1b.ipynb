```python
students = []
courses = []
marks = {}

def input_number_of_students():
    return int(input("Nhập số lượng sinh viên trong lớp: "))

def input_students_info(num_students):
    for i in range(num_students):
        print(f"\n--- Nhập thông tin sinh viên thứ {i + 1} ---")
        student_id = input("ID sinh viên: ")
        name = input("Họ và tên: ")
        dob = input("Ngày sinh (DD/MM/YYYY): ")
        students.append({
            'id': student_id,
            'name': name,
            'dob': dob
        })

def input_number_of_courses():
    return int(input("\nNhập số lượng môn học: "))

def input_courses_info(num_courses):
    for i in range(num_courses):
        print(f"\n--- Nhập thông tin môn học thứ {i + 1} ---")
        course_id = input("ID môn học: ")
        name = input("Tên môn học: ")
        courses.append({
            'id': course_id,
            'name': name
        })

def input_marks_for_course():
    if not courses or not students:
        print("Chưa có thông tin sinh viên hoặc môn học!")
        return

    print("\n--- Danh sách môn học hiện có ---")
    list_courses()
    selected_course_id = input("\nNhập ID môn học muốn nhập điểm: ")

    course_exists = any(c['id'] == selected_course_id for c in courses)
    if not course_exists:
        print("Không tìm thấy môn học có ID này!")
        return

    print(f"\n--- Nhập điểm cho môn học {selected_course_id} ---")
    for student in students:
        mark = float(input(f"Nhập điểm cho sinh viên {student['name']} (ID: {student['id']}): "))
        marks[(selected_course_id, student['id'])] = mark

def list_courses():
    print("\n================ DANH SÁCH MÔN HỌC ================")
    if not courses:
        print("Chưa có môn học nào.")
        return
    for c in courses:
        print(f"ID: {c['id']} | Tên môn: {c['name']}")

def list_students():
    print("\n================ DANH SÁCH SINH VIÊN ================")
    if not students:
        print("Chưa có sinh viên nào.")
        return
    for s in students:
        print(f"ID: {s['id']} | Tên: {s['name']} | Ngày sinh: {s['dob']}")

def show_marks_for_course():
    if not courses:
        print("Chưa có dữ liệu môn học!")
        return

    list_courses()
    course_id = input("\nNhập ID môn học cần xem điểm: ")

    print(f"\n================ BẢNG ĐIỂM MÔN {course_id} ================")
    found = False
    for student in students:
        key = (course_id, student['id'])
        if key in marks:
            print(f"ID: {student['id']} | Tên: {student['name']} | Điểm: {marks[key]}")
            found = True

    if not found:
        print("Chưa có điểm nào được nhập cho môn học này.")

def main():
    num_students = input_number_of_students()
    input_students_info(num_students)

    num_courses = input_number_of_courses()
    input_courses_info(num_courses)

    while True:
        print("\n================ MENU QUẢN LÝ ================")
        print("1. Nhập điểm cho môn học")
        print("2. Xem danh sách môn học")
        print("3. Xem danh sách sinh viên")
        print("4. Xem điểm của sinh viên theo môn học")
        print("0. Thoát chương trình")

        choice = input("Lựa chọn của bạn (0-4): ")

        if choice == '1':
            input_marks_for_course()
        elif choice == '2':
            list_courses()
        elif choice == '3':
            list_students()
        elif choice == '4':
            show_marks_for_course()
        elif choice == '0':
            print("Đã thoát chương trình.")
            break
        else:
            print("Lựa chọn không hợp lệ, vui lòng chọn lại!")

if __name__ == "__main__":
    main()
```
